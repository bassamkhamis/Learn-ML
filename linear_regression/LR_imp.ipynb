```python
import numpy as np
```

```python
# Simulate this equation y = 2x + 2, line equation
# X = [1, 3, 5, 7, 8] 
Y = np.array([4, 8, 12, 16, 18])
# augmantied vector, add 1 for bias
X = np.array([ [1,1], [1, 3], [1, 5], [1, 7], [1, 8] ])

# W = (X^t * X)^-1 * X^t * Ys
W = np.matmul(X.T, X)
W = np.linalg.inv(W)
W = np.matmul(W, X.T)
W = np.matmul(W, Y)
W
```

```python
# Add some noise
Y_noise = np.array([4 + 0.2, 8 - .1, 12, 16 + 0.3, 18 - 0.2])

W = np.matmul(X.T, X)
W = np.linalg.inv(W)
W = np.matmul(W, X.T)
W = np.matmul(W, Y_noise)
W
# result very close
```
